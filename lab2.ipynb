```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("Lab2-Transactions")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")
print(f"Spark {spark.version} — gotowy")
```

```python
df = spark.read.json("transactions_10k.jsonl")

print(f"Liczba rekordów: {df.count()}")
df.printSchema()
```

```python
df.show(10, truncate=False)
```

```python
from pyspark.sql.functions import to_timestamp, col

df = df.withColumn("timestamp", to_timestamp(col("timestamp"), "yyyy-MM-dd HH:mm:ss"))

df.printSchema()  # timestamp powinien być teraz 'timestamp (nullable = true)'
```

```python
from pyspark.sql.functions import count, sum as _sum, avg, round as _round

store_summary = (
    df.groupBy("store")
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
        _round(avg("amount"), 2).alias("srednia_PLN"),
    )
    .orderBy("store")
)
store_summary.show()
```

```python
from pyspark.sql.functions import min as _min, max as _max, sum as _sum

df.groupBy("category") \
  .agg(
      _sum("amount").alias("suma_PLN"),
      _min("amount").alias("min_PLN"),
      _max("amount").alias("max_PLN")
  ) \
  .orderBy("category") \
  .show()
```

```python
from pyspark.sql.functions import window

hourly = (
    df.groupBy(window("timestamp", "1 hour"))    # okno 1-godzinne
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
    )
    .orderBy("window")
)
hourly.show(truncate=False)
```

```python
(
    hourly
    .select(
        col("window.start").alias("od"),
        col("window.end").alias("do"),
        "liczba_tx",
        "suma_PLN",
    )
    .show(truncate=False)
)
```

```python
from pyspark.sql.functions import window

df.groupBy(window("timestamp", "30 minutes"), "store") \
  .agg(
      count("tx_id").alias("liczba_tx"),
      _round(_sum("amount"), 2).alias("suma_PLN")
  ) \
  .orderBy("window", "store") \
  .show(truncate=False)
```

```python
from pyspark.sql.functions import desc

# TWÓJ KOD - Zadanie 3.3
df.filter(col("store") == "Kraków") \
  .groupBy(window("timestamp", "1 hour")) \
  .agg(
      _round(_sum("amount"), 2).alias("suma_PLN")
  ) \
  .orderBy(desc("suma_PLN")) \
  .show(1, truncate=False) # show(1) pokaże tylko pierwszy (najwyższy) wynik
```

```python
sliding = (
    df.groupBy(window("timestamp", "1 hour", "30 minutes"))  # szerokość 1h, krok 30min
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
    )
    .select(
        col("window.start").alias("od"),
        col("window.end").alias("do"),
        "liczba_tx",
        "suma_PLN",
    )
    .orderBy("od")
)
sliding.show(truncate=False)
```

```python
tumbling_rows = (
    df.groupBy(window("timestamp", "1 hour"))
    .agg(count("tx_id"))
    .count()
)
sliding_rows = (
    df.groupBy(window("timestamp", "1 hour", "30 minutes"))
    .agg(count("tx_id"))
    .count()
)
print(f"Tumbling (1h):          {tumbling_rows} okien")
print(f"Sliding  (1h / 30min):  {sliding_rows} okien")

# Odpowiedz w komentarzu: dlaczego sliding ma więcej wierszy?
# TWOJA ODPOWIEDŹ:
# Okno 'sliding' generuje więcej wierszy wynikowych, ponieważ przedziały czasowe nakładają się na siebie. 
# W przypadku okna 1-godzinnego przesuwanego co 30 minut, jedna transakcja wpada do dwóch różnych okien naraz 
# (np. transakcja z 09:15 zostanie zaliczona do okna 08:30-09:30 oraz do okna 09:00-10:00). 
# Z tego powodu sumaryczna liczba unikalnych okien (i wygenerowanych wierszy) jest większa niż w oknach 'tumbling', 
# gdzie każde zdarzenie trafia dokładnie do jednego, nienakładającego się przedziału.
```
